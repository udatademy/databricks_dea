### Configurar acceso a los Datos mediante Unity Catalog
1. Usar el catalog "demo"
2. Crear y usar el schema "Raw"
3. Crear el volumen "data_skew"

#### 1. Usar el catalog "demo"

```sql
USE CATALOG demo;
```

#### 2. Crear y usar el schema "Raw"

```sql
CREATE SCHEMA IF NOT EXISTS demo.raw
    COMMENT 'Schema Raw';
```

```sql
USE SCHEMA raw;
```

#### 3. Crear el volumen "data_skew"

```sql
CREATE EXTERNAL VOLUME IF NOT EXISTS demo.raw.data_skew
    LOCATION 'abfss://demo@databricksdeasac.dfs.core.windows.net/raw/data_skew'
    COMMENT 'Volumen Data Skew';
```
