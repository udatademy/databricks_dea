### Optimización Predictiva (Predictive Optimization)
La Optimización Predictiva automatiza el mantenimiento de las tablas Delta administradas por Unity Catalog. Ejecuta operaciones como `OPTIMIZE`, `VACUUM` y `ANALYZE` cuando considera que son necesarias, ayudando a mejorar la organización de los archivos, gestionar el almacenamiento y mantener un buen rendimiento de las consultas, reduciendo la necesidad de realizar tareas de mantenimiento manualmente.
> 1. Validar que **`Predictive Optimization`** este habilitado en el **`Manage Account`**.
> 2. Crear el catalog **`empresa`** y habilitar **`Predictive Optimization`**
> 3. Describir de forma detallado el catalog **`empresa`**.
> 4. Crear el schema **`landing`** (desactivar Predictive Optimization).
> 5. Describir de forma detallado el schema **`landing`**
> 6. Crear la tabla **`employees`** en el schema **`landing`**.
> 7. Describir de forma detallado la tabla **`employees`**
> 8. Crear el schema **`processing`** (sin Predictive Optimization)
> 9. Describir de forma detallado el schema **`processing`**
> 10. Crear la tabla **`employees`** en el schema **`processing`**.
> 7. Describir de forma detallado la tabla **`employees`**

```sql
CREATE CATALOG IF NOT EXISTS empresa;
ALTER CATALOG empresa ENABLE PREDICTIVE OPTIMIZATION;
```

```sql
DESCRIBE CATALOG EXTENDED empresa;
```

```sql
CREATE SCHEMA IF NOT EXISTS empresa.landing;
ALTER SCHEMA empresa.landing DISABLE PREDICTIVE OPTIMIZATION;
```

```sql
DESCRIBE SCHEMA EXTENDED empresa.landing;
```

```sql
CREATE TABLE IF NOT EXISTS empresa.landing.employees (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    email STRING,
    department STRING,
    salary DECIMAL(10,2),
    hire_date DATE
);
```

```sql
DESCRIBE TABLE EXTENDED empresa.landing.employees;
```

```sql
CREATE SCHEMA IF NOT EXISTS empresa.processing;
```

```sql
DESCRIBE SCHEMA EXTENDED empresa.processing;
```

```sql
CREATE TABLE IF NOT EXISTS empresa.processing.employees (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    email STRING,
    department STRING,
    salary DECIMAL(10,2),
    hire_date DATE
);
```

```sql
DESCRIBE TABLE EXTENDED empresa.processing.employees;
```
